# 02 – Exploratory Data Analysis

Explore `data/processed/clean_dataset.csv` (or feature-engineered data if available).
Figures are saved under `reports/figures/`.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

ROOT = Path("..").resolve()
if not (ROOT / "data" / "processed").exists():
    ROOT = Path(".").resolve()

FIG = ROOT / "reports" / "figures"
FIG.mkdir(parents=True, exist_ok=True)

clean_path = ROOT / "data" / "processed" / "clean_dataset.csv"
df = pd.read_csv(clean_path, parse_dates=["date_of_record"])
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
df.head()
```

```python
print(df.isnull().sum())
df["rainfall"].describe()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["rainfall"], bins=50)
plt.title("Rainfall Distribution")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(FIG / "rainfall_distribution.png", dpi=150)
plt.show()
```

```python
if "month" in df.columns:
    month_order = [
        "January", "February", "March", "April", "May", "June",
        "July", "August", "September", "October", "November", "December",
    ]
    monthly = df.groupby("month")["rainfall"].mean()
    monthly = monthly.reindex([m for m in month_order if m in monthly.index])
    monthly.plot(kind="bar", figsize=(10, 5))
    plt.title("Average Monthly Rainfall")
    plt.ylabel("Average Rainfall (mm)")
    plt.xlabel("Month")
    plt.tight_layout()
    plt.savefig(FIG / "monthly_rainfall.png", dpi=150)
    plt.show()
else:
    print("month column not present (already dropped in feature engineering).")
```

```python
if "season" in df.columns:
    season = df.groupby("season")["rainfall"].mean()
    season.plot(kind="bar", figsize=(8, 5))
    plt.title("Average Season-wise Rainfall")
    plt.ylabel("Average Rainfall (mm)")
    plt.xlabel("Season")
    plt.tight_layout()
    plt.savefig(FIG / "season_rainfall.png", dpi=150)
    plt.show()
else:
    print("season column not present (already dropped in feature engineering).")
```

```python
corr = df.select_dtypes(include="number").corr()
plt.figure(figsize=(10, 8))
plt.imshow(corr)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.savefig(FIG / "correlation_matrix.png", dpi=150)
plt.show()
corr
```
