# Exploratory Data Analysis
This notebook explores the cleaned rainfall dataset and prepares it for modeling.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/clean_dataset.csv")
df.head()
```

```python
print(df.shape)
print(df.columns)
df.isnull().sum()
df["rainfall"].describe()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["rainfall"], bins=50)
plt.title("rainfall Distribution")
plt.xlabel("rainfall (mm)")
plt.ylabel("Frequency")
plt.show()
```

```python
monthly = df.groupby("month")["rainfall"].mean()

monthly.plot(kind="bar", figsize=(10,5))
plt.ylabel("Average Rainfall")
plt.show()
```

```python
season = df.groupby("season")["rainfall"].mean()

season.plot(kind="bar")
plt.show()
```

```python
season = df.groupby("season")["rainfall"].mean()

season.plot(kind="bar")
plt.show()
```

```python
state = df.groupby("state")["rainfall"].mean()

state.sort_values().tail(20).plot(kind="barh")
plt.show()
```

```python
import matplotlib.pyplot as plt

corr = df.select_dtypes(include="number").corr()

plt.figure(figsize=(10,8))
plt.imshow(corr)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.show()
```
