# Exploratory Data Analysis
This notebook explores the cleaned rainfall dataset and prepares it for modeling.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

project_root = Path.cwd().resolve()
for candidate in [project_root, project_root.parent, project_root.parent.parent]:
    if (candidate / "src").exists():
        project_root = candidate
        break

sys.path.insert(0, str(project_root))

clean_data_path = project_root / 'data' / 'clean_dataset.csv'

if clean_data_path.exists():
    df = pd.read_csv(clean_data_path)
    print(df.head())
else:
    print(f"Clean dataset not found at {clean_data_path}. Run the preprocessing notebook first.")
```

```python
if 'df' in locals():
    print('Shape:', df.shape)
    print('\nMissing values:\n', df.isnull().sum())
    print('\nDuplicate rows:', df.duplicated().sum())
else:
    print('Load the dataset first.')
```

```python
if 'df' in locals() and 'Rainfall' in df.columns:
    plt.figure(figsize=(10, 4))
    sns.histplot(df['Rainfall'].dropna(), bins=30, kde=True)
    plt.title('Rainfall Distribution')
    plt.xlabel('Rainfall')
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()
    print(df['Rainfall'].describe())
else:
    print('Rainfall column not found or dataset not loaded.')
```

```python
if 'df' in locals():
    plt.figure(figsize=(10, 6))
    corr = df.select_dtypes(include=['number']).corr(numeric_only=True)
    sns.heatmap(corr, annot=False, cmap='coolwarm')
    plt.title('Feature Correlation Heatmap')
    plt.tight_layout()
    plt.show()
else:
    print('Load the dataset first.')
```
