# 05 – Sequence Generation

Build contiguous 30-day sequences for the LSTM baseline.

**Locked rules**
- Group by `station_id`
- No gap filling — contiguous calendar days only
- `X` = weather + DOY cyclical features (no lat/lon/elevation)
- Split by target date: Train ≤2022, Val 2023, Test 2024–2025-02-10
- MinMaxScaler fit on train only

```python
# Run the project-root script (keeps logic in one place)
import subprocess
import sys
from pathlib import Path

root = Path("..").resolve()
if not (root / "generate_sequences.py").exists():
    root = Path(".").resolve()

result = subprocess.run(
    [sys.executable, str(root / "generate_sequences.py")],
    cwd=str(root),
    capture_output=True,
    text=True,
)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
    raise SystemExit(result.returncode)
```
