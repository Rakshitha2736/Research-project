# 05 – Sequence Generation

Build contiguous 30-day sequences for the LSTM baseline.

**Locked rules**
- Group by `station_id`
- No gap filling — contiguous calendar days only
- Primary (v2): X includes past `rainfall` → 8 features
- Split by target date: Train ≤2022, Val 2023, Test 2024–2025-02-10
- MinMaxScaler fit on train only

This notebook runs `generate_sequences_v2.py` (preferred). Use `generate_sequences.py` only for the older 7-feature ablation.

```python
import subprocess
import sys
from pathlib import Path

root = Path("..").resolve()
if not (root / "generate_sequences_v2.py").exists():
    root = Path(".").resolve()

script = root / "generate_sequences_v2.py"
result = subprocess.run(
    [sys.executable, str(script)],
    cwd=str(root),
    capture_output=True,
    text=True,
)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
    raise SystemExit(result.returncode)
```
